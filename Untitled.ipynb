```python
import talib

# RSI 14 (best standard short-term momentum indicator)
df['RSI_14'] = talib.RSI(df['Close'], timeperiod=14)

# EMA 12 (fast EMA, great for short-term trend)
df['EMA_12'] = talib.EMA(df['Close'], timeperiod=12)
```

```python
tickers = {
    "Reliance": "RELIANCE.NS",
    "HDFC": "HDFCBANK.NS",
    "ICICI": "ICICIBANK.NS",
    "Infosys": "INFY.NS",
    "TCS": "TCS.NS",
    "SP500": "^GSPC",
    "Nasdaq": "^IXIC",
    "DowJones": "^DJI",

    "CrudeOil": "BZ=F",
    "Gold": "GC=F",
    "USDINR": "INR=X"
}
Asian_ticker={
      "Nikkei": "^N225",
    "HangSeng": "^HSI",
    "Kospi": "^KS11",
}
```

```python
for name, code in tickers.items():
    temp = yf.download(code, start="2010-01-01")
    temp[name] = ((temp['Close'] - temp['Open']) / temp['Open']) * 100
    temp=temp[[name]]
    if isinstance(temp.columns, pd.MultiIndex):
        temp.columns = temp.columns.droplevel(1)
    df = df.join(temp, how="left")






```

```python
for name,code in Asian_ticker.items():
   temp1 = yf.download(code,start="2010-01-01")
   if isinstance(temp1.columns, pd.MultiIndex):
      temp1.columns = temp1.columns.droplevel(1)
   temp1['prev_close']=temp1['Close'].shift(1)
   temp1[name] = (temp1['Open'] - temp1['prev_close']) / temp1['prev_close'] * 100
   temp1=temp1[[name]]
   df = df.join(temp1, how="left")
```

```python
df = df[df['Volume'] > 0]
```

```python
import numpy as np

df['Gapup'] = np.where(df['target_value'] > df['Close'], 1, 0)

```

```python

split = int(len(df) * 0.7)

train = df[:split]
test  = df[split:]

features = ['Open', 'Close', 'High', 'Low', 'Volume', 'percentage_change', 'RSI_14',
       'EMA_12', 'Reliance', 'HDFC', 'ICICI', 'Infosys', 'TCS', 'SP500',
       'Nasdaq', 'DowJones', 'Nikkei', 'HangSeng', 'Kospi', 'CrudeOil', 'Gold',
       'USDINR']

X_train = train[features]
X_test  = test[features]

y_train = train['Gapup']
y_test  = test['Gapup']


```

```python
X_train
```

```python
from sklearn.ensemble import RandomForestClassifier, VotingClassifier
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
from sklearn.model_selection import TimeSeriesSplit, cross_val_score,GridSearchCV,RandomizedSearchCV
```

```python
param_grid = {

    'rf__n_estimators': [100, 200, 300, 500],
    'rf__max_depth': [None, 5, 10, 15, 20],
    'rf__min_samples_split': [2, 5, 10],
    'rf__min_samples_leaf': [1, 2, 4],
    'rf__max_features': ['sqrt', 'log2', 0.5, 0.8],
    'rf__bootstrap': [True, False],


    'xgb__n_estimators': [100, 200, 300, 500],
    'xgb__max_depth': [3, 5, 6, 8, 10],
    'xgb__learning_rate': [0.01, 0.05, 0.1, 0.2],
    'xgb__subsample': [0.6, 0.8, 1.0],
    'xgb__colsample_bytree': [0.6, 0.8, 1.0],
    'xgb__gamma': [0, 1, 5],
    'xgb__min_child_weight': [1, 3, 5],
    'xgb__reg_lambda': [0.1, 1, 5, 10],
    'xgb__reg_alpha': [0, 0.1, 1]
}
```

```python

tscv = TimeSeriesSplit(n_splits=8)

rf = RandomForestClassifier(n_estimators=150, max_depth=40, random_state=18, n_jobs=-1)
xgb = XGBClassifier(n_estimators=300, max_depth=50, learning_rate=0.01, subsample=0.9, colsample_bytree=0.9, eval_metric='logloss', random_state=42)

ensemble = VotingClassifier(estimators=[('rf', rf), ('xgb', xgb)], voting='soft', weights=[2,1])



ensemble.fit(X_train, y_train)

```

```python
y_pred = ensemble.predict(X_test)
print(accuracy_score(y_test, y_pred))
print(precision_score(y_test, y_pred))
print(recall_score(y_test, y_pred))
print(f1_score(y_test, y_pred))
```
